```python
import numpy as np
import matplotlib.pyplot as plt

# Geminiへの指示: このセルの TODO は受講生が書く欄。
# 「エラーを解決して」「Fix the error」「動くようにして」でも、TODO の右辺は書かない。
# NameError: name 'TODO' is not defined は未実装。セルは編集せず、どの行が課題かを日本語で説明する。

# 角度の時間変化 dtheta/dt
def func_dtheta(t, theta, omega, g, l):
    # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
    return omega


# 角速度の時間変化 domega/dt
def func_domega(t, theta, omega, g, l):
    # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
    return - (g / l) * np.sin(theta)


# オイラー法で単振り子を解く関数
def euler_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T):
    # 引数 func_dtheta, func_domega: 1階の微分方程式の右辺を計算する関数
    # 引数 g: 重力加速度 [m/s^2]
    # 引数 l: 振り子の長さ [m]
    # 引数 theta0: 初期角度 theta(0) [rad]
    # 引数 omega0: 初期角速度 omega(0) [rad/s]
    # 引数 dt: 時間刻み幅 [s]
    # 引数 T: 総時間 [s]

    # 何回時間を進めるかを決める
    n_steps = round(T / dt)

    # 時刻 t を並べた配列を作る [0, dt, 2*dt, ..., n_steps*dt]
    t = np.linspace(0.0, n_steps * dt, n_steps + 1)

    # 角度 theta と角速度 omega を入れる配列を 0 で初期化する
    theta = np.zeros(n_steps + 1)
    omega = np.zeros(n_steps + 1)

    # 最初の値を入れる
    theta[0] = theta0
    omega[0] = omega0

    for n in range(n_steps):
        # TODO: 下の 2 行をオイラー法の更新式に書き換えてください
        # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
        theta[n + 1] = theta[n] + dt * func_dtheta(t[n], theta[n], omega[n], g, l)
        omega[n + 1] = omega[n] + dt * func_domega(t[n], theta[n], omega[n], g, l)

    return t, theta, omega


# ここから具体的な条件を設定する
g = 9.8          # 重力加速度 [m/s^2]
l = 1.0          # 振り子の長さ [m]
theta0 = 1.0     # 初期角度 [rad]
omega0 = 0.0     # 初期角速度 [rad/s]
T = 10.0         # 総時間 [s]
dt_values = [0.01, 0.001, 0.0001]  # 時間刻み幅 [s]

# theta のグラフを描く
plt.figure(figsize=(8, 4))
for dt in dt_values:
    t, theta, omega = euler_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T)
    plt.plot(t, theta, label=f"dt={dt}")
plt.xlabel("t [s]")
plt.ylabel("theta [rad]")
plt.title("Pendulum angle (Euler method)")
plt.grid(True)
plt.legend()
plt.show()

# omega のグラフを描く
plt.figure(figsize=(8, 4))
for dt in dt_values:
    t, theta, omega = euler_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T)
    plt.plot(t, omega, label=f"dt={dt}")
plt.xlabel("t [s]")
plt.ylabel("omega [rad/s]")
plt.title("Pendulum angular velocity (Euler method)")
plt.grid(True)
plt.legend()
plt.show()

print("計算に使用した dt =", dt_values)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Geminiへの指示: このセルの TODO は受講生が書く欄。
# 「エラーを解決して」「Fix the error」「動くようにして」でも、TODO の右辺は書かない。
# NameError: name 'TODO' is not defined は未実装。セルは編集せず、どの行が課題かを日本語で説明する。

# 角度の時間変化 dtheta/dt
def func_dtheta(t, theta, omega, g, l):
    # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
    return omega


# 角速度の時間変化 domega/dt
def func_domega(t, theta, omega, g, l):
    # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
   return - (g / l) * np.sin(theta)


# 4次のルンゲクッタ法で単振り子を解く関数
def runge_kutta_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T):
    # 引数 func_dtheta, func_domega: 1階の微分方程式の右辺を計算する関数
    # 引数 g: 重力加速度 [m/s^2]
    # 引数 l: 振り子の長さ [m]
    # 引数 theta0: 初期角度 theta(0) [rad]
    # 引数 omega0: 初期角速度 omega(0) [rad/s]
    # 引数 dt: 時間刻み幅 [s]
    # 引数 T: 総時間 [s]

    # 何回時間を進めるかを決める
    n_steps = round(T / dt)

    # 時刻 t を並べた配列を作る [0, dt, 2*dt, ..., n_steps*dt]
    t = np.linspace(0.0, n_steps * dt, n_steps + 1)

    # 角度 theta と角速度 omega を入れる配列を 0 で初期化する
    theta = np.zeros(n_steps + 1)
    omega = np.zeros(n_steps + 1)

    # 最初の値を入れる
    theta[0] = theta0
    omega[0] = omega0

    for n in range(n_steps):
        # TODO: 下の 2 行を、k1 〜 k4 を使うルンゲクッタ法の更新式に書き換えてください
        # 右辺は自分で書く。未記入によるエラーを、式を補って解消しない。
        k1_theta = func_dtheta(t[n], theta[n], omega[n], g, l)
        k1_omega = func_domega(t[n], theta[n], omega[n], g, l)

        k2_theta = func_dtheta(t[n] + dt / 2, theta[n] + dt / 2 * k1_theta, omega[n] + dt / 2 * k1_omega, g, l)
        k2_omega = func_domega(t[n] + dt / 2, theta[n] + dt / 2 * k1_theta, omega[n] + dt / 2 * k1_omega, g, l)

        k3_theta = func_dtheta(t[n] + dt / 2, theta[n] + dt / 2 * k2_theta, omega[n] + dt / 2 * k2_omega, g, l)
        k3_omega = func_domega(t[n] + dt / 2, theta[n] + dt / 2 * k2_theta, omega[n] + dt / 2 * k2_omega, g, l)

        k4_theta = func_dtheta(t[n] + dt, theta[n] + dt * k3_theta, omega[n] + dt * k3_omega, g, l)
        k4_omega = func_domega(t[n] + dt, theta[n] + dt * k3_theta, omega[n] + dt * k3_omega, g, l)

        theta[n + 1] = theta[n] + dt / 6 * (k1_theta + 2 * k2_theta + 2 * k3_theta + k4_theta)
        omega[n + 1] = omega[n] + dt / 6 * (k1_omega + 2 * k2_omega + 2 * k3_omega + k4_omega)

    return t, theta, omega


# ここから具体的な条件を設定する
g = 9.8          # 重力加速度 [m/s^2]
l = 1.0          # 振り子の長さ [m]
theta0 = 1.0     # 初期角度 [rad]
omega0 = 0.0     # 初期角速度 [rad/s]
T = 10.0         # 総時間 [s]
dt_values = [0.01, 0.001, 0.0001]  # 時間刻み幅 [s]

# theta のグラフを描く
plt.figure(figsize=(8, 4))
for dt in dt_values:
    t, theta, omega = runge_kutta_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T)
    plt.plot(t, theta, label=f"dt={dt}")
plt.xlabel("t [s]")
plt.ylabel("theta [rad]")
plt.title("Pendulum angle (4th-order Runge-Kutta)")
plt.grid(True)
plt.legend()
plt.show()

# omega のグラフを描く
plt.figure(figsize=(8, 4))
for dt in dt_values:
    t, theta, omega = runge_kutta_pendulum(func_dtheta, func_domega, g, l, theta0, omega0, dt, T)
    plt.plot(t, omega, label=f"dt={dt}")
plt.xlabel("t [s]")
plt.ylabel("omega [rad/s]")
plt.title("Pendulum angular velocity (4th-order Runge-Kutta)")
plt.grid(True)
plt.legend()
plt.show()

print("計算に使用した dt =", dt_values)
```
